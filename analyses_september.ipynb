```python
## CHECKING IF ALL THE VARIABLES HAVE BEEN GROUPED
df_cols = df.columns.tolist()

# Flatten grouped_variables dictionary into a list of all variables
grouped_variables_list = [v for sublist in grouped_variables.values() for v in sublist]

# Find variables in df_cols that are not in grouped_variables_list
cols_missing = [col for col in df_cols if col not in grouped_variables_list]

#print(cols_missing)

df_cols_missing = [var for var in grouped_variables_list if var not in df_cols]
print(df_cols_missing)
```

```python
## CHECKING REDUNDANCY 
# Threshold for high correlation
threshold = 0.95

# Store results
redundant_pairs = []

for domain, variables in grouped_variables.items():
    # Keep only variables present in your df
    domain_vars = [v for v in variables if v in df.columns]
    if len(domain_vars) < 2:
        continue  # Skip if not enough variables

    # Select numeric variables
    domain_df = df[domain_vars].select_dtypes(include=[np.number])

    # Skip if no numeric variables
    if domain_df.shape[1] < 2:
        continue

    # Compute correlation matrix
    corr_matrix = domain_df.corr(method='pearson')

    # Extract upper triangle without diagonal
    mask = np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
    corr_pairs = corr_matrix.where(mask).stack().reset_index()
    corr_pairs.columns = ['var1', 'var2', 'correlation']

    # Filter by threshold
    high_corr = corr_pairs[abs(corr_pairs['correlation']) > threshold].copy()
    high_corr['domain'] = domain

    redundant_pairs.append(high_corr)

# Combine all results
if redundant_pairs:
    redundant_df = pd.concat(redundant_pairs, ignore_index=True)
    print("Highly correlated variable pairs within domains:")
    print(redundant_df.sort_values(by='correlation', ascending=False))
else:
    print("No high correlations found within domains.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# ---------- FUNCTION ----------
def plot_corr(ax, df, cols, labels):
    present = [c for c in cols if c in df.columns]
    data = df[present].dropna()
    corr = data.corr().rename(index=labels, columns=labels)

    mask = np.triu(np.ones_like(corr, dtype=bool), k=1)

    sns.heatmap(
        corr,
        annot=True,
        cmap='Reds',
        fmt='.2f',
        mask=mask,
        square=True,
        linewidths=.5,
        cbar=False,
        ax=ax
    )
    



# ---------- DATA ----------
# Panel A: EAA
epi_clocks_eaa = ['EAA_LEVINE', 'EAA_GRIMAGE', 'EAA_DUNEDINMPOA', 'EAA_HORVATH', 'EAA_HANNUM']
labels_eaa = {
    'EAA_LEVINE': 'PhenoAge',
    'EAA_GRIMAGE': 'GrimAge',
    'EAA_DUNEDINMPOA': 'DunedinPoAm',
    'EAA_HORVATH': 'Horvath',
    'EAA_HANNUM': 'Hannum'
}

# Panel B: DNAm age + chronological age
epi_clocks_age = ['CHRON_AGE', 'LEVINE_DNAMAGE', 'DNAMGRIMAGE', 'MPOA', 'HORVATH_DNAMAGE', 'HANNUM_DNAMAGE']
labels_age = {
    'CHRON_AGE': 'Age',
    'LEVINE_DNAMAGE': 'PhenoAge',
    'DNAMGRIMAGE': 'GrimAge',
    'MPOA': 'DunedinPoAm',
    'HORVATH_DNAMAGE': 'Horvath',
    'HANNUM_DNAMAGE': 'Hannum'
}

# ---------- FIGURE ----------
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

plot_corr(axes[0], df, epi_clocks_age, labels_age)
plot_corr(axes[1], df, epi_clocks_eaa, labels_eaa)
# AFTER plotting each heatmap

axes[0].text(
    -0.15, 1.05, "A",
    transform=axes[0].transAxes,
    fontsize=14,
    fontweight='bold',
    va='top',
    ha='left'
)

axes[1].text(
    -0.15, 1.05, "B",
    transform=axes[1].transAxes,
    fontsize=14,
    fontweight='bold',
    va='top',
    ha='left'
)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt

epi_clocks = ['EAA_LEVINE', 'EAA_GRIMAGE', 'EAA_DUNEDINMPOA', 'EAA_HORVATH', 'EAA_HANNUM']
age = df['CHRON_AGE'].to_numpy()

for col in epi_clocks:
    if col in df.columns:
        y = df[col].to_numpy()
        X = sm.add_constant(age)
        m = sm.OLS(y, X).fit()
        resid = m.resid
        fitted = m.fittedvalues

        # 1) Residuals vs Age
        plt.figure(figsize=(6,4))
        plt.scatter(age, resid, alpha=0.5, s=12)
        plt.axhline(0, color='k', lw=1)
        plt.xlabel('Chronological Age'); plt.ylabel(f'{col} residuals')
        plt.title(f'{col}: Residuals vs Age')
        plt.tight_layout(); plt.show()

        # 2) Scale–location: sqrt|resid| vs fitted
        plt.figure(figsize=(6,4))
        plt.scatter(fitted, np.sqrt(np.abs(resid)), alpha=0.5, s=12)
        plt.xlabel('Fitted EpiAge'); plt.ylabel('sqrt(|residuals|)')
        plt.title(f'{col}: Scale–Location (variance check)')
        plt.tight_layout(); plt.show()

        # 3) (Optional) Age-binned variance
        bins = np.quantile(age, [0, .2, .4, .6, .8, 1.0])
        cats = np.digitize(age, bins[1:-1], right=True)
        var_by_bin = [np.var(resid[cats==k], ddof=1) for k in range(len(bins)-1)]
        print(f'{col}: variance by age bin -> {var_by_bin}')
```

```python
from statsmodels.stats.diagnostic import het_breuschpagan, het_white

for col in epi_clocks:
    if col in df.columns:
        y = df[col].to_numpy()
        X = sm.add_constant(age)
        m = sm.OLS(y, X).fit()
        # Explanatory vars for the variance: constant + Age (and optionally Age^2)
        exog = m.model.exog  # [const, Age]
        bp_stat, bp_p, _, _ = het_breuschpagan(m.resid, exog)
        w_stat, w_p, _, _  = het_white(m.resid, exog)
        print(f'{col}: BP p={bp_p:.4g} | White p={w_p:.4g}')
```
