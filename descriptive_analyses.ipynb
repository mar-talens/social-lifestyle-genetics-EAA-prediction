```python
# ...existing code...
import importlib
import importlib.util
import os
import pandas as pd
import sys

# ...existing code...
def load_variable_descriptions(source="/Users/martalens/Desktop/ML_EAA/data/metadata/variables_dictionary.csv"):
    if isinstance(source, dict):
        return source
    if isinstance(source, str) and os.path.isfile(source):
        ext = os.path.splitext(source)[1].lower()
        if ext == ".csv":
            df = pd.read_csv(source)
            cols = {c.lower(): c for c in df.columns}
            # variable and description candidates (include your raw_var)
            var_cands = ("variable","var","name","feature","raw_var","rawvar")
            desc_cands = ("description","desc","label","help","long_name")
            var_col = next((cols[c] for c in var_cands if c in cols), None)
            desc_col = next((cols[c] for c in desc_cands if c in cols), None)
            if var_col is None or desc_col is None:
                raise ValueError("CSV must contain variable and description columns. Found: " + ", ".join(df.columns))
            return df[[var_col, desc_col]].dropna(subset=[var_col]).set_index(var_col)[desc_col].astype(str).to_dict()
        if ext == ".py":
            path = os.path.abspath(source)
            try:
                spec = importlib.util.spec_from_file_location("variables_dictionary_from_file", path)
                mod = importlib.util.module_from_spec(spec)
                spec.loader.exec_module(mod)
            except Exception as e:
                print(f"Failed to load python file {path}: {e}")
                return {}
        else:
            # unknown file type -> try to import as module name fallback
            try:
                mod = importlib.import_module(source)
            except Exception as e:
                print(f"Unable to handle file {source}: {e}")
                return {}
        # try common dict names in module
        candidates = ["raw_var"]
        for name in candidates:
            if hasattr(mod, name):
                val = getattr(mod, name)
                if isinstance(val, dict):
                    return val
        # fallback: first dict attribute
        for k, v in vars(mod).items():
            if isinstance(v, dict):
                return v
        print("No dict found in module/file.")
        return {}

    # if looks like a module import name
    if isinstance(source, str):
        try:
            mod = importlib.import_module(source)
        except Exception as e:
            print(f"Failed to import {source}: {e}")
            return {}
        candidates = ["description"]
        for name in candidates:
            if hasattr(mod, name):
                val = getattr(mod, name)
                if isinstance(val, dict):
                    return val
        for k, v in vars(mod).items():
            if isinstance(v, dict):
                return v
        print("No dict found in imported module.")
        return {}

    return {}

def print_grouped_with_descriptions(grouped_variables, descriptions=None, save_csv=None):
    if descriptions is None:
        descriptions = load_variable_descriptions()
    rows = []
    for grp, varlist in grouped_variables.items():
        if isinstance(varlist, str):
            vars_iter = [v.strip() for v in varlist.split(",") if v.strip()]
        else:
            vars_iter = list(varlist)
        vars_sorted = sorted(vars_iter, key=lambda s: (s or "").lower())
        print(f"\n--- {grp} ({len(vars_sorted)} variables) ---")
        for v in vars_sorted:
            desc = None
            if isinstance(descriptions, dict):
                desc = descriptions.get(v)
                if desc is None:
                    desc = descriptions.get(v.upper())
                if desc is None:
                    desc = descriptions.get(v.lower())
            if desc is None:
                desc = "<no description found>"
            print(f"{v}: {desc}")
            rows.append({"group": grp, "variable": v, "description": desc})
    if save_csv:
        pd.DataFrame(rows).to_csv(save_csv, index=False)
        print(f"\nSaved descriptions by group to {save_csv}")
    return pd.DataFrame(rows)

# Example: load from CSV (change path if needed)
descriptions = load_variable_descriptions("/Users/martalens/Desktop/ML_EAA/data/metadata/variables_dictionary.csv")
# Or from the existing .py
# descriptions = load_variable_descriptions("/Users/martalens/Desktop/ML_EAA/data/metadata/variables_dictionary.py")

# Run and save
df_descriptions_by_group = print_grouped_with_descriptions(grouped_variables,
                                                          load_variable_descriptions("/Users/martalens/Desktop/ML_EAA/data/metadata/variables_dictionary.csv"),
                                                          save_csv="/Users/martalens/Desktop/ML_EAA/output/variable_descriptions_by_group.csv")
# ...existing code...
```

```python

import pandas as pd
import numpy as np
from scipy.stats import spearmanr

# --- paths ---
shap_path = "/Users/martalens/Desktop/ML_EAA/output/nov_analyses/XGB/Regressor/shap_long_EAA_GRIMAGE.csv"
perm_path = "/Users/martalens/Desktop/ML_EAA/output/nov_analyses/XGB/Regressor/permutation_feature_standard.csv"

CLOCK = "EAA_GRIMAGE"   # change to "EAA_DUNEDINMPOA" if needed
TOPK  = 15

# --- read ---
shap = pd.read_csv(shap_path)
perm = pd.read_csv(perm_path)

# --- aggregate SHAP: mean |shap_value| per feature (within selected clock) ---
shap_c = shap.loc[shap["clock"].eq(CLOCK)].copy()
shap_agg = (
    shap_c.assign(abs_shap=shap_c["shap_value"].abs())
          .groupby("variable", as_index=True)["abs_shap"].mean()
          .rename("mean_abs_shap")
)

# --- aggregate permutation: mean drop in R² per feature across folds (within clock) ---
perm_c = perm.loc[perm["clock"].eq(CLOCK)].copy()
perm_agg = (
    perm_c.groupby("feature", as_index=True)["mean_drop"].mean()
          .rename("mean_drop_r2")
)

# --- align on common features ---
df = (
    pd.concat([shap_agg, perm_agg], axis=1)
      .rename_axis("feature")
      .dropna()              # keep intersection only
)

# --- concordance metrics ---
rho, pval = spearmanr(df["mean_abs_shap"], df["mean_drop_r2"])

top_shap = set(df.sort_values("mean_abs_shap", ascending=False).head(TOPK).index)
top_perm = set(df.sort_values("mean_drop_r2",   ascending=False).head(TOPK).index)
overlap  = len(top_shap & top_perm)
overlap_pct = 100 * overlap / TOPK

print(f"Clock: {CLOCK}")
print(f"Spearman ρ = {rho:.3f} (p = {pval:.3g})")
print(f"Top-{TOPK} overlap = {overlap}/{TOPK} ({overlap_pct:.1f}%)")

# --- optional: normalized columns & save table for SI ---
df = df.assign(
    shap_prop = df["mean_abs_shap"] / df["mean_abs_shap"].sum(),
    perm_prop = df["mean_drop_r2"]  / df["mean_drop_r2"].sum()
).sort_values("mean_abs_shap", ascending=False)

out_csv = f"S1_shap_perm_concordance_{CLOCK}.csv"
df.reset_index().to_csv(out_csv, index=False)
print("Saved:", out_csv)
print(df.head(15))

K = 15

# top-K by each metric
top_shap_idx = df.sort_values("mean_abs_shap", ascending=False).head(K).index
top_perm_idx = df.sort_values("mean_drop_r2",   ascending=False).head(K).index

# union of the two sets
top_union = top_shap_idx.union(top_perm_idx)

df_top = df.loc[top_union].copy()

rho_top, pval_top = spearmanr(df_top["mean_abs_shap"], df_top["mean_drop_r2"])
print(f"Spearman on top-{K} union: ρ = {rho_top:.3f}, p = {pval_top:.3g}")
```

```python
grouped_variables = {
    'adult adversity and life-course events': [
        'ALCOHOL_DRUG_FAM','ATTACK','CHLDIED','COMBAT','CURRENT_MARRIAGE_YRS_RAND','DENCARE','DENLOAN','DENPROM','DISASTER',
        'DISCRIMINATION','DISMJOB','DRUGOTH','EVER_AGE_DISCRIMINATION','EVER_ANCESTRY_DISCRIMINATION','EVER_APPAREANCE_DISCRIMINATION',
        'EVER_DISABILITY_DISCRIMINATION','EVER_FINANCIAL_DISCRIMINATION','EVER_GENDER_DISCRIMINATION','EVER_LIVED_AS_COUPLE',
        'EVER_ORIENTATION_DISCRIMINATION','EVER_RACE_DISCRIMINATION','EVER_RELIGION_DISCRIMINATION','EVER_WEIGHT_DISCRIMINATION',
        'FINANCIAL_STRAIN_ELDERLY','FINANCIAL_STRAIN_YRSLF','FRAUD_5Y_AGG','HOMELESS_AGG','HOUSE_AT40_1.0','HOUSE_AT40_2.0',
        'HOUSE_AT40_3.0','HOUSE_AT40_7.0','HOUSE_FIRSTJOB_1.0','HOUSE_FIRSTJOB_2.0','HOUSE_FIRSTJOB_3.0','HOUSE_FIRSTJOB_4.0',
        'HOUSE_FIRSTJOB_5.0','HOUSING_PROB_YOURSELF','ILLOTH','ILLSELF','JAIL_AGG','LH15','LH19','LH1B3','LH1B4','LH3A','LH4B','LH4C',
        'LH4D','LH4E','LH51','LIVING_AT40_ALONE','LIVING_AT40_BIOCHILD','LIVING_AT40_GRANDP','LIVING_AT40_INLAWS','LIVING_AT40_NONREL',
        'LIVING_AT40_OTHREL','LIVING_AT40_PARENTS','LIVING_AT40_PARTNER','LIVING_AT40_SIBS','LIVING_AT40_STEPCHILD','LIVING_AT40_CHILDREN_UNSPEC',
        'LIVING_ATFULLJOB_ALONE','LIVING_ATFULLJOB_BIOCHILD','LIVING_ATFULLJOB_GRANDP','LIVING_ATFULLJOB_INLAWS','LIVING_ATFULLJOB_MILITARY',
        'LIVING_ATFULLJOB_NONREL','LIVING_ATFULLJOB_OTHREL','LIVING_ATFULLJOB_PARENTS','LIVING_ATFULLJOB_PARTNER','LIVING_ATFULLJOB_STEPCHILD',
        'LIVING_ATFULLJOB_SIBS','LONGEST_MARRIAGE_YRS_RAND','LOST_JOB_5Y_AGG','MOVED_5Y_AGG','NEIGHBORHOOD_COHESION','NEIGHBORHOOD_DISORDER',
        'NEVER_MARRIED','NHIREDJOB','NUMBER_DIVORCES','NUMBER_LIVING_SIBLINGS','NUMBER_MARRIAGES','NUMBER_WIDOWED','PHY_EMOT_PROB_FAM',
        'PREVMOV','RELATIONSHIP_PROB','ROBBED_5Y_AGG','SICK_FAM','UNEMPLOYED_5Y_AGG','UNFPOLICE','VETERAN_RAND','WORK_DIFF_YRSLF', 'NO_LHMS_SUPP'
    ],
    'demographic': [
        'BIRTH_PLACE_1.0','BIRTH_PLACE_10.0','BIRTH_PLACE_11.0','BIRTH_PLACE_2.0','BIRTH_PLACE_3.0','BIRTH_PLACE_4.0','BIRTH_PLACE_5.0', 'BIRTH_PLACE_6.0','BIRTH_PLACE_7.0','BIRTH_PLACE_8.0','BIRTH_PLACE_9.0','BIRTHYR','GENDER','HISPANIC_BINARY','IMMGYEAR','LH1', 'NO_LHMS_SUPP',
        'MARITAL_STATUS_DIVORCED','MARITAL_STATUS_MARRIED','MARITAL_STATUS_NEVER_MARRIED','MARITAL_STATUS_PARTNERED',
        'MARITAL_STATUS_WIDOWED','RAEVBRN_CAT','RELIGION_1.0','RELIGION_2.0','RELIGION_3.0','RELIGION_4.0','RELIGION_5.0',
        'SIBLINGS_BINARY','USBORN', 'AFRICAN_AMERICAN_BINARY', 'WHITE_BINARY', 'NO_ETHNICITY_BINARY'
    ],
    'early-life conditions and family background': [
        'ATTENMO','CHMISSCH','DAD_AGE_RAND','DAD_ALIVE_RAND','DRKDRUG','DRUG','EFFMO','FAEDUC','FAMFIN','FATHER_DISABLED','FAUNEM','FMFINH', 'FJOB_ARMEDFORCES','FJOB_BLUECOLLAR', 'FJOB_SERVICE', 'FJOB_WHITECOLLAR', 'MOTHER_AGE_RAND',        'MOTHER_ALIVE_RAND', 'NUM_SCHOOLS', 'RELFA_NOTAPPLY',  'HOUSE_AT10_1.0','HOUSE_AT10_2.0','HOUSE_AT10_3.0','HOUSE_AT10_4.0','LH11','LH1B1','LH1B2','LH23','LH2A','LH2B','LH2C','LH2D','LH2E', 'NO_LHMS_SUPP',
        'LH2G','LH2H','LH2I','LH30','LH31A','LH31B','LH31C','LH31D','LH31E','LH31F','LH33','LH7','LH8','LIVEGPAR','LIVING_AT10_BIOFATH','LIVING_AT10_BIOMOTH','LIVING_AT10_GRANDP','LIVING_AT10_NONREL','LIVING_AT10_OTHREL','LIVING_AT10_SIBLINGS','LIVING_AT10_STEPFATH', 'LIVING_AT10_STEPMOTH','MOEDUC','MOVFIN','MOWORK','MULTILINGUAL_HOME','NEVER_LIVED_FATHER','NEVER_LIVED_MOTHER','PARSMOKE','PHYABUSE','PRIVATE_SCHOOL','RELWFA','RELWMO','RELWMO_NOTAPPLY','RTHLTHCH','SCHLOVER','SPOKE_ENGLISH_HOME','TEACHMO','TRPOLICE'
    ],
    'education, job and socioeconomic status': [
        'CHRONIC_UNEMPLOYMENT','CURRENTLY_PENSION','CURRENTLY_WORKING','DEGREE','FINANCIAL_CONTROL_CHANGE','HH_INCOME_DROPS','HH_MEAN_INCOME', 'HH_POVERTY_THRESHOLD_CAT',
        'HH_PERSISTENT_LOW_INCOME','HH_UNEMPLOYED_5Y_AGG','IND_PUBLIC','IND_SECONDARY','IND_TERTIARY','IND_UNKNOWN','LH13','LH34','LH38', 'NO_LHMS_SUPP',
        'LH39A','LH39B','LH39C','LH39D','LH40A','LH40B','LH40C','LH40D','LONGEST_JOB_DURATION','NUM_UNEMPLOYMENT','NUM_UNI','OCC_ARMEDFORCES',
        'OCC_BLUECOLLAR','OCC_SERVICE','OCC_WHITECOLLAR','OCC_UNKNOWN','PRIVATE_UNI','RAEDUC','SCHLYRS'
    ],
    'genetics': [
        'PC1_5A','PC1_5B','PC1_5C','PC1_5D','PC1_5E','PC6_10A','PC6_10B','PC6_10C','PC6_10D','PC6_10E','PGI_AB_BROAD17','PGI_ADHD_PGC17',
        'PGI_AFBC_SOCGEN16','PGI_AI_GSCAN19','PGI_ALC_PGC18','PGI_ANXFS_ANGST16','PGI_AUTISM_PGC17','PGI_BIP_PGC11','PGI_BMI2_GIANT18',
        'PGI_BUN_CKDGEN19','PGI_BUNTE_CKDGEN19','PGI_CAD_CARDIOGRAM11','PGI_CANNABIS_ICC18','PGI_CKD_CKDGEN19','PGI_CKDTE_CKDGEN19',
        'PGI_CRP_CHARGE22','PGI_DEPSYMP_SSGAC16','PGI_DPW_GSCAN19','PGI_EDU3_W23_SSGAC18','PGI_EGFR_CKDGEN19','PGI_EGFRTE_CKDGEN19',
        'PGI_EXTRAVERSION_GPC16','PGI_GENCOG2_CHARGE18','PGI_GWALZNA_PGC21','PGI_HBA1CEA_MAGIC17','PGI_HDL_GLGC13','PGI_HTN_COGNET17',
        'PGI_LDL_GLGC13','PGI_MENARCHE_REPROGEN17','PGI_MENOPAUSEREPROGEN21','PGI_MI_CARDIOGRAM15','PGI_NEBC_SOCGEN16',
        'PGI_NEUROTICISM_SSGAC16','PGI_OCD_IOCDF17','PGI_PTSDC_PGC18','PGI_SC_GSCAN19','PGI_SCZ_PGC14','PGI_SI_GSCAN19',
        'PGI_T2DALL_DIAGRAM24','PGI_TC_GLGC13','PGI_TG_GLGC13','PGI_WC_GIANT15','PGI_WELLBEING_SSGAC16','PGI_WHR_GIANT15','PGI_XDISORDER_PGC13'
    ],
    'lifestyle and health behaviors': [
        'BINGE_LIFECOURSE','BMI','CHSMOKE','CURRENTLY_DRINKS_RAND','CURRENTLY_SMOKING_RAND','EVER_SMOKED_RAND','DAYS_WEEK_DRINKING_RAND', 'NO_LHMS_SUPP', 
        'DRINKS_PER_DAY_RAND','DRINKS_PER_WEEK_RAND','HEALTH_PROB_YRSLF','LH58A','LH58B','LH58C','LH59A','LH59B','LH59C',
        'PHYS_ACTIVITY_MILD','PHYS_ACTIVITY_MODERATE','PHYS_ACTIVITY_VIGOROUS'
    ],
    'social network and support': [
        'CHILDREN_NEG_SUPPORT','CHILDREN_POS_SUPPORT','CLOSE_CHILDREN','CLOSE_FAMILY','CLOSE_FRIENDS','CONTACT_CHILDREN','CONTACT_FAMILY',
        'CONTACT_FRIENDS','FAMILY_NEG_SUPPORT','FAMILY_POS_SUPPORT','FRIENDS_NEG_SUPPORT','FRIENDS_POS_SUPPORT','HAS_CHILDREN','HAS_FAMILY',
        'HAS_FRIENDS','LIVES_WITH_PARTNER','LONELINESS','RELATIONSHIP_SPOUSE','RELATIVES_NEIGHBORHOOD','SPOUSE_NEG_SUPPORT','SPOUSE_POS_SUPPORT'
    ],
    "smoking": ["CURRENTLY_SMOKING_RAND","EVER_SMOKED_RAND", "CHSMOKE"],

    'lifestyle without smoking': [
        'BINGE_LIFECOURSE','BMI', 'CURRENTLY_DRINKS_RAND', 'DAYS_WEEK_DRINKING_RAND', 'NO_LHMS_SUPP', 
        'DRINKS_PER_DAY_RAND','DRINKS_PER_WEEK_RAND','HEALTH_PROB_YRSLF','LH58A','LH58B','LH58C','LH59A','LH59B','LH59C',
        'PHYS_ACTIVITY_MILD','PHYS_ACTIVITY_MODERATE','PHYS_ACTIVITY_VIGOROUS']
}
# ...existing code...

# Summary of missing values per variable-group (number and range)
import pandas as pd
import numpy as np

def missing_by_group(df, grouped_variables, save_csv="/Users/martalens/Desktop/ML_EAA/output/missing_by_group.csv"):
    rows = []
    group_summaries = []
    N = len(df)
    for gname, varlist in grouped_variables.items():
        # ensure varlist is iterable and flatten accidental single string entries
        if isinstance(varlist, str):
            vars_iter = [v.strip() for v in varlist.split(",") if v.strip()]
        else:
            vars_iter = list(varlist)
        # normalize names (strip)
        vars_iter = [v.strip() for v in vars_iter]
        # keep only variables present in df
        present = [v for v in vars_iter if v in df.columns]
        missing_vars = [v for v in vars_iter if v not in df.columns]
        if missing_vars:
            print(f"Group '{gname}': {len(missing_vars)} variables not found in df (first 5): {missing_vars[:5]}")
        counts = []
        for v in present:
            miss = int(df[v].isna().sum())
            pct  = 100 * miss / N
            rows.append({"group": gname, "variable": v, "missing_count": miss, "missing_pct": pct, "n_obs": N - miss})
            counts.append(miss)
        if counts:
            group_summaries.append({
                "group": gname,
                "n_vars": len(present),
                "min_missing": int(np.min(counts)),
                "max_missing": int(np.max(counts)),
                "median_missing": float(np.median(counts)),
                "mean_missing": float(np.mean(counts))
            })
        else:
            group_summaries.append({
                "group": gname,
                "n_vars": 0,
                "min_missing": np.nan,
                "max_missing": np.nan,
                "median_missing": np.nan,
                "mean_missing": np.nan
            })
    df_long = pd.DataFrame(rows).sort_values(["group","missing_count"], ascending=[True, False]).reset_index(drop=True)
    df_group = pd.DataFrame(group_summaries).sort_values("group").reset_index(drop=True)
    # save detailed table
    df_long.to_csv(save_csv, index=False)
    print(f"Saved detailed missing table to {save_csv}")
    return df_long, df_group

# Call the function
d_long, d_group = missing_by_group(df, grouped_variables)

# Quick prints
print("\nPer-group summary (min / max / mean missing per variable):")
print(d_group[['group','n_vars','min_missing','max_missing','mean_missing']].to_string(index=False))

print("\nTop 10 most-missing variables overall:")
print(d_long.sort_values("missing_count", ascending=False).head(10).to_string(index=False))
```

```python
# ...existing code...
import pandas as pd
import numpy as np

def missing_pct_all_vars(df, grouped_variables,
                         save_long="/Users/martalens/Desktop/ML_EAA/output/missing_by_group_allvars.csv",
                         save_group="/Users/martalens/Desktop/ML_EAA/output/missing_group_summary_allvars.csv"):
    N = len(df)
    rows = []
    for gname, varlist in grouped_variables.items():
        if isinstance(varlist, str):
            vars_iter = [v.strip() for v in varlist.split(",") if v.strip()]
        else:
            vars_iter = list(varlist)
        for v in vars_iter:
            if v in df.columns:
                miss = int(df[v].isna().sum())
                pct  = 100 * miss / N
                n_obs = N - miss
                present = True
            else:
                miss = np.nan
                pct  = np.nan
                n_obs = 0
                present = False
            rows.append({
                "group": gname,
                "variable": v,
                "present_in_df": present,
                "missing_count": miss if present else np.nan,
                "missing_pct": pct if present else np.nan,
                "n_obs": n_obs
            })
    df_long_all = pd.DataFrame(rows)

    # per-group summary (only over variables actually present in df)
    df_present = df_long_all[df_long_all["present_in_df"]].copy()
    df_group = (df_present.groupby("group")["missing_pct"]
                .agg(n_present_vars="count",
                     min_missing_pct="min",
                     max_missing_pct="max",
                     median_missing_pct="median",
                     mean_missing_pct="mean")
                .reset_index()
               )

    # compute group-level total missing % (all variables together)
    group_totals = (df_present.groupby("group")["missing_count"]
                    .agg(total_missing_count="sum")
                    .reset_index())
    df_group = df_group.merge(group_totals, on="group", how="left")
    df_group["total_possible_cells"] = df_group["n_present_vars"] * N
    df_group["group_total_missing_pct"] = 100 * df_group["total_missing_count"] / df_group["total_possible_cells"]

    # also report number of listed variables missing from df per group
    missing_counts = (df_long_all.loc[~df_long_all["present_in_df"]]
                      .groupby("group")["variable"].count()
                      .rename("n_listed_vars_not_in_df")
                      .reset_index())
    df_group = df_group.merge(missing_counts, on="group", how="left").fillna({"n_listed_vars_not_in_df":0})

    # overall summary across all listed & present variables
    overall_total_missing = df_present["missing_count"].sum()
    overall_n_vars_present = df_present["variable"].nunique()
    overall_total_cells = overall_n_vars_present * N if overall_n_vars_present > 0 else np.nan
    overall_total_missing_pct = (100 * overall_total_missing / overall_total_cells) if overall_total_cells and not np.isnan(overall_total_cells) else np.nan

    # dataset-level stats across variables (min/max/mean/median of per-variable missing %)
    pct_series = df_present["missing_pct"].dropna()
    pct_stats = {
        "vars_min_missing_pct": float(pct_series.min()) if not pct_series.empty else np.nan,
        "vars_max_missing_pct": float(pct_series.max()) if not pct_series.empty else np.nan,
        "vars_mean_missing_pct": float(pct_series.mean()) if not pct_series.empty else np.nan,
        "vars_median_missing_pct": float(pct_series.median()) if not pct_series.empty else np.nan,
        "overall_total_missing_pct": float(overall_total_missing_pct) if not np.isnan(overall_total_missing_pct) else np.nan
    }

    overall = {
        "overall_n_vars_present": int(overall_n_vars_present) if not np.isnan(overall_n_vars_present) else 0,
        "overall_total_missing_count": int(overall_total_missing) if not np.isnan(overall_total_missing) else 0,
        "overall_total_cells": int(overall_total_cells) if not np.isnan(overall_total_cells) else 0,
        **pct_stats
    }

    # save
    df_long_all.to_csv(save_long, index=False)
    df_group.to_csv(save_group, index=False)
    return df_long_all, df_group, overall

# call it
d_all, d_group, overall = missing_pct_all_vars(df, grouped_variables)

# quick prints
print("\nPer-group summary (missing % stats over variables present in df + group-total pct):")
print(d_group[["group","n_present_vars","min_missing_pct","max_missing_pct","mean_missing_pct","group_total_missing_pct","n_listed_vars_not_in_df"]].to_string(index=False))

print("\nTop 20 variables by missing % (only variables present in df):")
print(d_all[d_all["present_in_df"]].sort_values("missing_pct", ascending=False).head(20)[["group","variable","missing_pct","n_obs"]].to_string(index=False))

print("\nDataset-level missing % stats across all listed & present variables:")
print(overall)
# ...existing code...
```
